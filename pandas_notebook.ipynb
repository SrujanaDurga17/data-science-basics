```python
df["age"].fillna(df["age"].mean(),inplace=True)
#replacing the missing values in the "age" column with the mean of that column. This will help to ensure that we have a complete dataset for analysis, and it can also help to reduce bias that may be introduced by simply dropping rows with missing values.
# inplace is to reflect the applied changes to df
```

```python
#to find total fare
total_fare =int(df["fare"].sum())
print(total_fare)
```

```python
#class wise fare
classwise_fare=df.groupby("pclass")["fare"].sum()
print(classwise_fare)
```

```python
#class wise count
class_count=df.groupby("pclass")["name"].count()
print(class_count)
```

```python
import pandas as pd
classwise_fare_df= pd.DataFrame(classwise_fare)
class_count_df=pd.DataFrame(class_count)
print(classwise_fare_df)
```

```python
df_merge= pd.merge(class_count_df,classwise_fare_df,on="pclass")
df_merge.rename(columns={"name":"Total_People","fare":"Total_Fare"},inplace=True)
# renaming to columns
print(df_merge)
```
